```python
# Calculate correlation matrix
corr_matrix = features.corr()

# Get upper triangle to avoid duplicate pairs
upper_triangle = corr_matrix.where(
    np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
)

# Find highly correlated features
high_corr = (
    upper_triangle.stack()
    .reset_index()
)

high_corr.columns = ['Feature 1', 'Feature 2', 'Correlation']

high_corr = high_corr[
    high_corr['Correlation'].abs() > 0.90
]

high_corr['Correlation'] = high_corr['Correlation'].round(3)

high_corr = high_corr.sort_values(
    by='Correlation',
    key=abs,
    ascending=False
)

display(high_corr)
```

## Outliers Detection

```python
plt.figure(figsize=(22, 25))

for i, column in enumerate(features.columns, 1):

    plt.subplot(6, 5, i)

    sns.boxplot(
        y=features[column]
    )

    plt.title(column, fontsize=9)

plt.tight_layout()
plt.show()
```

```python
outlier_results = []

for column in features.columns:

    Q1 = features[column].quantile(0.25)
    Q3 = features[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = features[
        (features[column] < lower_bound) |
        (features[column] > upper_bound)
    ][column]

    outlier_results.append({
        'Feature': column,
        'Outliers Count': len(outliers),
        'Outliers Percentage': round(
            len(outliers) / len(features) * 100, 2
        )
    })

outlier_df = pd.DataFrame(outlier_results)

outlier_df = outlier_df.sort_values(
    by='Outliers Count',
    ascending=False
)

display(outlier_df)
```

```python
df = df.drop(columns=['Unnamed: 0'])

print("New Dataset Shape:", df.shape)

df.head()
```

```python
X = df.drop(columns=['y'])

y = df['y']

print("Features Shape:", X.shape)
print("Target Shape:", y.shape)
```

```python
y = y.map({
    'B': 0,
    'M': 1
})

print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Features:", X_train.shape)
print("Testing Features:", X_test.shape)

print("\nTraining Target Distribution:")
print(y_train.value_counts(normalize=True) * 100)

print("\nTesting Target Distribution:")
print(y_test.value_counts(normalize=True) * 100)
```

```python
from sklearn.preprocessing import RobustScaler
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
```

```python
lr_model = Pipeline([
    ('scaler', RobustScaler()),
    ('model', LogisticRegression(max_iter=5000))
])

knn_model = Pipeline([
    ('scaler', RobustScaler()),
    ('model', KNeighborsClassifier())
])

svm_model = Pipeline([
    ('scaler', RobustScaler()),
    ('model', SVC(probability=True, random_state=42))
])
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier

# Decision Tree
dt_model = DecisionTreeClassifier(random_state=42)

# Random Forest
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Gradient Boosting
gb_model = GradientBoostingClassifier(random_state=42)
```

```python
models = {
    'Logistic Regression': lr_model,
    'KNN': knn_model,
    'SVM': svm_model,
    'Decision Tree': dt_model,
    'Random Forest': rf_model,
    'Gradient Boosting': gb_model
}
```

```python
from sklearn.model_selection import StratifiedKFold, cross_validate
import pandas as pd

# Stratified 5-Fold Cross Validation
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# Evaluation Metrics
scoring = {
    'accuracy': 'accuracy',
    'precision': 'precision',
    'recall': 'recall',
    'f1': 'f1'
}

results = []

for name, model in models.items():

    scores = cross_validate(
        model,
        X_train,
        y_train,
        cv=cv,
        scoring=scoring
    )

    results.append({
        'Model': name,
        'Accuracy': scores['test_accuracy'].mean(),
        'Precision': scores['test_precision'].mean(),
        'Recall': scores['test_recall'].mean(),
        'F1-Score': scores['test_f1'].mean()
    })

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by='F1-Score',
    ascending=False
)

results_df
```

```python
from sklearn.ensemble import VotingClassifier

# Hard Voting
hard_voting = VotingClassifier(
    estimators=[
        ('lr', lr_model),
        ('svm', svm_model),
        ('knn', knn_model)
    ],
    voting='hard'
)

# Soft Voting
soft_voting = VotingClassifier(
    estimators=[
        ('lr', lr_model),
        ('svm', svm_model),
        ('knn', knn_model)
    ],
    voting='soft'
)
```

```python
voting_models = {
    'Hard Voting': hard_voting,
    'Soft Voting': soft_voting
}

voting_results = []

for name, model in voting_models.items():

    scores = cross_validate(
        model,
        X_train,
        y_train,
        cv=cv,
        scoring=scoring
    )

    voting_results.append({
        'Model': name,
        'Accuracy': scores['test_accuracy'].mean(),
        'Precision': scores['test_precision'].mean(),
        'Recall': scores['test_recall'].mean(),
        'F1-Score': scores['test_f1'].mean()
    })

voting_df = pd.DataFrame(voting_results)

voting_df
```

## Bagging Classifier

```python
from sklearn.ensemble import BaggingClassifier

bagging_model = BaggingClassifier(
    estimator=DecisionTreeClassifier(random_state=42),
    n_estimators=100,
    max_samples=0.8,
    bootstrap=True,
    random_state=42,
    n_jobs=-1
)
```

```python
bagging_scores = cross_validate(
    bagging_model,
    X_train,
    y_train,
    cv=cv,
    scoring=scoring
)

bagging_result = pd.DataFrame([{
    'Model': 'Bagging Classifier',
    'Accuracy': bagging_scores['test_accuracy'].mean(),
    'Precision': bagging_scores['test_precision'].mean(),
    'Recall': bagging_scores['test_recall'].mean(),
    'F1-Score': bagging_scores['test_f1'].mean()
}])

bagging_result
```

## Stacking Ensemble

```python
from sklearn.ensemble import StackingClassifier

stacking_A = StackingClassifier(
    estimators=[
        ('svm', svm_model),
        ('knn', knn_model),
        ('dt', dt_model)
    ],
    final_estimator=LogisticRegression(max_iter=5000),
    cv=StratifiedKFold(
        n_splits=5,
        shuffle=True,
        random_state=42
    ),
    stack_method='auto',
    n_jobs=-1
)
```

```python
stacking_A_scores = cross_validate(
    stacking_A,
    X_train,
    y_train,
    cv=cv,
    scoring=scoring
)

stacking_A_result = pd.DataFrame([{
    'Model': 'Stacking Experiment A',
    'Accuracy': stacking_A_scores['test_accuracy'].mean(),
    'Precision': stacking_A_scores['test_precision'].mean(),
    'Recall': stacking_A_scores['test_recall'].mean(),
    'F1-Score': stacking_A_scores['test_f1'].mean()
}])

stacking_A_result
```

## Experiment B

```python
stacking_B = StackingClassifier(
    estimators=[
        ('rf', rf_model),
        ('gb', gb_model),
        ('svm', svm_model)
    ],
    final_estimator=LogisticRegression(max_iter=5000),
    cv=StratifiedKFold(
        n_splits=5,
        shuffle=True,
        random_state=42
    ),
    stack_method='auto',
    n_jobs=-1
)
```

```python
stacking_B_scores = cross_validate(
    stacking_B,
    X_train,
    y_train,
    cv=cv,
    scoring=scoring
)

stacking_B_result = pd.DataFrame([{
    'Model': 'Stacking Experiment B',
    'Accuracy': stacking_B_scores['test_accuracy'].mean(),
    'Precision': stacking_B_scores['test_precision'].mean(),
    'Recall': stacking_B_scores['test_recall'].mean(),
    'F1-Score': stacking_B_scores['test_f1'].mean()
}])

stacking_B_result
```

## Meta Model

```python
stacking_B_rf_meta = StackingClassifier(
    estimators=[
        ('rf', rf_model),
        ('gb', gb_model),
        ('svm', svm_model)
    ],
    final_estimator=RandomForestClassifier(
        n_estimators=100,
        random_state=42
    ),
    cv=StratifiedKFold(
        n_splits=5,
        shuffle=True,
        random_state=42
    ),
    stack_method='auto',
    n_jobs=-1
)
```

```python
stacking_B_rf_scores = cross_validate(
    stacking_B_rf_meta,
    X_train,
    y_train,
    cv=cv,
    scoring=scoring
)

stacking_B_rf_result = pd.DataFrame([{
    'Model': 'Stacking B - Random Forest Meta',
    'Accuracy': stacking_B_rf_scores['test_accuracy'].mean(),
    'Precision': stacking_B_rf_scores['test_precision'].mean(),
    'Recall': stacking_B_rf_scores['test_recall'].mean(),
    'F1-Score': stacking_B_rf_scores['test_f1'].mean()
}])

stacking_B_rf_result
```

## Hyperparameter Tuning

```python
from sklearn.model_selection import GridSearchCV

stacking_param_grid = {
    'final_estimator__C': [0.01, 0.1, 1, 10, 100],
    'final_estimator__class_weight': [None, 'balanced'],
    'final_estimator__solver': ['lbfgs']
}

stacking_grid = GridSearchCV(
    estimator=stacking_B,
    param_grid=stacking_param_grid,
    cv=cv,
    scoring='f1',
    n_jobs=-1,
    verbose=1
)

stacking_grid.fit(X_train, y_train)
```

```python
print("Best Parameters:")
print(stacking_grid.best_params_)

print("\nBest Cross Validation F1-Score:")
print(stacking_grid.best_score_)
```

```python
from sklearn.base import clone

seeds = [0, 21, 42, 100, 123]

seed_results = []

for seed in seeds:

    model = clone(stacking_grid.best_estimator_)

    model.set_params(
        cv=StratifiedKFold(
            n_splits=5,
            shuffle=True,
            random_state=seed
        )
    )

    scores = cross_validate(
        model,
        X_train,
        y_train,
        cv=StratifiedKFold(
            n_splits=5,
            shuffle=True,
            random_state=seed
        ),
        scoring=scoring,
        n_jobs=-1
    )

    seed_results.append({
        'Random State': seed,
        'Accuracy': scores['test_accuracy'].mean(),
        'Precision': scores['test_precision'].mean(),
        'Recall': scores['test_recall'].mean(),
        'F1-Score': scores['test_f1'].mean()
    })

seed_results_df = pd.DataFrame(seed_results)

seed_results_df
```

```python
seed_summary = seed_results_df[
    ['Accuracy', 'Precision', 'Recall', 'F1-Score']
].agg(['mean', 'std'])

seed_summary
```

```python
import time

timing_models = {
    'Logistic Regression': lr_model,
    'Random Forest': rf_model,
    'Bagging Classifier': bagging_model,
    'Stacking Experiment B (Tuned)': stacking_grid.best_estimator_
}

timing_results = []

for name, model in timing_models.items():
    m = clone(model)
    start = time.perf_counter()
    m.fit(X_train, y_train)
    elapsed = time.perf_counter() - start
    timing_results.append({
        'Model': name,
        'Training Time (seconds)': round(elapsed, 3)
    })

timing_df = pd.DataFrame(timing_results)
timing_df
```

## Test Set

```python
from sklearn.base import clone

# Create the best model
final_model = clone(stacking_grid.best_estimator_)

# Train on training data only
final_model.fit(X_train, y_train)

# Predict on unseen test data
y_pred = final_model.predict(X_test)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

print("Final Test Results")

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1-Score:", f1_score(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(
    y_test,
    y_pred,
    target_names=['Benign', 'Malignant']
))
```

## Discussion


The experimental results show that ensemble learning methods can achieve strong performance on the Breast Cancer Wisconsin Diagnostic Dataset.

Among all evaluated models, Stacking Experiment B achieved the highest cross-validation F1-score of 96.75%, with an accuracy of 97.58% and a recall of 96.47%.

Logistic Regression also achieved excellent results, with an F1-score of 96.71%, demonstrating that a relatively simple individual classifier can perform competitively with more complex ensemble methods.

The comparison between Stacking Experiments A and B showed that the choice of base learners affects the overall performance. Experiment B achieved a higher recall and F1-score, while Experiment A achieved higher precision.

Replacing the Logistic Regression meta-model with Random Forest reduced the performance of Stacking Experiment B, indicating that a more complex meta-model does not necessarily improve the ensemble.

The Bagging Classifier improved the performance of the individual Decision Tree, increasing its F1-score from 89.62% to 93.99%.

Random seed experiments showed some variation in model performance, with a mean F1-score of 95.53% and a standard deviation of approximately 1.00%.

On the independent test set, the selected Stacking model achieved 96.49% accuracy, 100% precision, 90.48% recall, and 95.00% F1-score. Its confusion matrix showed 72 true negatives, 38 true positives, 4 false negatives, and no false positives.

Although the final test results are strong, the four false-negative cases highlight the importance of improving malignant-case detection. The model is intended for educational experimentation and not for clinical diagnosis.

**Does Stacking always outperform individual models?**
No. Stacking Experiment B achieved the best cross-validation F1-score (96.75%), but Logistic Regression alone reached 96.71%, which is essentially equal. In these results, Stacking did not beat the best individual model by a meaningful margin.

**Does combining diverse models lead to better performance?**
Experiment B combines three different model families (Random Forest, Gradient Boosting, and SVM) and achieved the best recall and F1-score. This is consistent with the idea that diverse models can complement each other, because their errors may be less correlated. However, this comparison alone does not isolate diversity as the cause, because the two experiments also differ in which individual algorithms were used.

**How does the choice of base models affect Stacking performance?**
It changes which metric improves. Experiment A (SVM, KNN, Decision Tree) achieved higher precision, while Experiment B achieved higher recall and F1-score. The base models should be chosen to be individually strong and complementary to each other.

**How does changing the meta-model affect the final predictions?**
Replacing Logistic Regression with Random Forest as the meta-model reduced performance in this experiment. One possible explanation is that a more complex meta-model can overfit the small set of meta-features, but this result alone does not prove the cause. What it shows is that a more complex meta-model is not automatically better.

**Is Stacking computationally more expensive than individual models?**
In the timing run above, the tuned Stacking model took much longer to train than the individual models we compared, because it trains every base model with internal cross-validation and then trains the meta-model on top. The exact numbers come from a single run on one machine, but the extra work is structural: Stacking always trains several base models plus a meta-model.

**Does Stacking improve all evaluation metrics or only specific ones?**
Only specific ones. Stacking Experiment B improved recall and F1-score, but not precision; Experiment A and some individual models achieved higher precision.

**Under what circumstances would Stacking be a better choice than other ensemble techniques?**
Stacking is a reasonable choice when the base models are strong and make different kinds of errors, when there is enough training data to produce reliable out-of-fold predictions, and when the extra computation time is acceptable. When the base models make very similar errors, the expected gain over simpler ensembles like Voting or Bagging is smaller.


## Conclusion

This project explored and compared individual machine learning classifiers and ensemble learning techniques for breast cancer classification.

Six individual classifiers were evaluated, followed by Voting, Bagging, and Stacking methods. Stacking Experiment B, using Random Forest, Gradient Boosting, and SVM as base learners with Logistic Regression as the meta-model, achieved the best cross-validation F1-score of 96.75%.

After hyperparameter tuning, the selected model achieved an F1-score of 95.00% on the independent test set, with an accuracy of 96.49%.

The results demonstrate the effectiveness of ensemble learning while also showing that model selection, validation strategy, and evaluation metrics play important roles in building reliable classification models.
